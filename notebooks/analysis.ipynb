## 2. Test du modèle Waterfall de base

Le modèle Waterfall comprend :
- Une file d'attente pour l'exécution des tests (K serveurs)
- Une file d'attente pour l'envoi des résultats (1 serveur)

### Configuration du modèle

```python
# random.seed(42)

arrival_rate = 0.8  # taux d'arrivée (Poisson)
k_tests = 2  # nombre de serveurs pour l'exécution
mu_tests = 0.5  # taux de service pour l'exécution
cap_tests = 50  # capacité de la file d'exécution
mu_result = 2.0  # taux de service pour l'envoi des résultats
cap_result = 20  # capacité de la file de résultats
duration = 100  # durée de simulation
```

### Exécution de la simulation

```python
sim = WaterfallSimulation(
    arrival_rate=arrival_rate,
    k_tests=k_tests,
    mu_tests=mu_tests,
    cap_tests=cap_tests,
    mu_result=mu_result,
    cap_result=cap_result,
    verbose=True,
)

print("Starting the Waterfall Simulation...")
sim.run(duration)
sim.print_report()
```
